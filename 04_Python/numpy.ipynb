Rearranging arrays

```python
#horizontally stack arrays
v1 = np.ones((2,3))
v2 = np.full((2,3),5)
np.hstack([v1,v2])
```

```python
#vertically stack arrays
k =np.vstack([v2,v1])
print(k)
```

```python
#horizontal spliting   - column wise
np.hsplit(k,3)
```

```python
#vertical splitting - row wise
np.vsplit(k,4)
```
