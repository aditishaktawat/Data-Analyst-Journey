Filtering

```python
#1st way
filt = df['Last_Name'] == 'Shaktawat'   #Returns a series of true false booleans
```

```python
# df[filt]
# or
# df.loc[filt]
```

```python
df.loc[filt, 'Email']   #return email where last name is 'Shaktawat'
```

```python
# 2nd way
df[df['Last_Name'] == 'Shaktawat']
```

```python
# AND filter
df[(df['First_Name'] == 'Aditi') & (df['Email'] == 'abc@email.com')]
```

```python
filt = (df['Last_Name'] == 'Shaktawat') & (df['Email'] == 'abc@email.com')
df.loc[filt]
```

```python
# OR condition
df[(df['Last_Name'] == 'Sin') | (df['Email'] == 'abc@email.com')]
```

```python
filt = (df['Last_Name'] == 'Sin') | (df['Email'] == 'abc@email.com')
df.loc[filt, 'First_Name']
```

```python
# NOT 
df.loc[~filt]   #opposite of filtering condition
```

APPLY

```python
# 1. APPLY with series - apply changes to each value of series

df['email'].apply(len)
```

```python
#cretaing a func for converting the case of email
def email_upper(email):
    return email.upper()
df['email'].apply(email_upper)
#apply to the column
df['email'] = df['email'].apply(email_upper)
```

```python
df
```

```python
# Or using LAMBDA func
df['email'] = df['email'].apply(lambda x: x.lower())
df
```

```python
# 2. Apply with dataframe - apply func to every series in dataframe
df.apply(len)
```

```python
df.apply(pd.Series.max)
```

```python
df.apply(lambda x: x.min())
```
