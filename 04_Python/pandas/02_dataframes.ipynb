```python
#Dictionary with single key-value pair
person = {
    "First_Name" : "Aditi",
    "Last_Name" : "Shaktawat",
    "Email" : "abc@email.com"
}
```

```python
#Dictionary with multiple values in key using list
person = {
    "First_Name" : ["Aditi", "Joe", "Anu"],
    "Last_Name" : ["Shaktawat", "Sin", "Dat"],
    "Email" : ["abc@email.com", "av@email.com", "fgees@email.com"]
}
```

```python
person['Email']
```

Creating a dataframe

```python
import pandas as pd
```

```python
df = pd.DataFrame(person)
```

```python
df
```

Series - Return single column

```python
#Accesing a singlr column in dataframe
df.Email 
#OR 
df['Email']    
```

```python
#Accesing Multiple columns - dataframe
df[['First_Name', 'Email']]
```

```python
#Return all columns
df.columns
```

Selecting Rows

```python
#Return all rows
# 1. Using iloc- integer location
df.iloc[[0,1]]
```

```python
df.iloc[[1,2],2]  #Return 2nd column rows with index 1,2
```

```python
#2. Using loc - search by labels
df.loc[[1,2], ['Email', 'First_Name']]
```

Filtering

```python
#1st way
filt = df['Last_Name'] == 'Shaktawat'   #Returns a series of true false booleans
```

```python
# df[filt]
# or
# df.loc[filt]
```

```python
df.loc[filt, 'Email']   #return email where last name is 'Shaktawat'
```

```python
# 2nd way
df[df['Last_Name'] == 'Shaktawat']
```

```python
# AND filter
df[(df['First_Name'] == 'Aditi') & (df['Email'] == 'abc@email.com')]
```

```python
filt = (df['Last_Name'] == 'Shaktawat') & (df['Email'] == 'abc@email.com')
df.loc[filt]
```

```python
# OR condition
df[(df['Last_Name'] == 'Sin') | (df['Email'] == 'abc@email.com')]
```

```python
filt = (df['Last_Name'] == 'Sin') | (df['Email'] == 'abc@email.com')
df.loc[filt, 'First_Name']
```

```python
# NOT 
df.loc[~filt]   #opposite of filtering condition
```
