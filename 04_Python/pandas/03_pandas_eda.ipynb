What % of people from each country knows python

```python
total_respondents= df['Country'].value_counts()
total_respondents
```

```python

python_respondants = df.groupby(['Country'])['LanguageHaveWorkedWith'].apply( lambda x: x.str.contains('Python').sum())
python_respondants
```

```python
#creating new dataframe for these 2 columns
python_df = pd.concat([total_respondents, python_respondants],axis='columns')
python_df
```

```python
 #Renaming columns
python_df.rename(columns={ 'count': 'Total_Respon' , 'LanguageHaveWorkedWith' : 'Knows_Python'},inplace= True)
python_df
```

```python
# Calculating percentage
python_df['PerctKnowsPython'] = (python_df['Knows_Python'] / python_df[ 'Total_Respon']) * 100
python_df
```

```python
# Sort results
python_df.sort_values(by='PerctKnowsPython' , ascending= False, inplace=True)
```
