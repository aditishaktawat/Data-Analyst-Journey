Filtering

```python
# Return age, country, years exp where exp is greater than 10 years
years_exp = (df['YearsCode'] > '10')
df.loc[years_exp, ['YearsCode', 'Age' ,'Country']]
```

```python
lang_known = df['LanguageHaveWorkedWith'] == 'Python'
df.loc[lang_known, 'LanguageHaveWorkedWith']
```

```python
# If countries in list, return age
Countries = ['India', 'Belgium','Germany', 'UK']
filt = df['Country'].isin(Countries)
df.loc[filt,'Country']
```

```python
# String Operations

#Filter where lang known has 'python' present in them

lang_present = df['LanguageHaveWorkedWith'].str.contains('Python', na=False)
df.loc[lang_present, 'LanguageHaveWorkedWith']
```
