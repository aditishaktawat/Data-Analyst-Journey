Convert strg to date format

```python
df['Date'] = pd.to_datetime(df['Date'], format='%b %d, %Y')
```

```python
df.loc[0,'Date'].day_name()
```

```python
# Date method on entire series
df['Date'].dt.day_name()
```

```python
# Creating new column from date
df['DayOfWeek'] = df['Date'].dt.day_name()
```

```python
df
```

Earliest and most recent dates

```python
df['Date'].min()
```

```python
df['Date'].max()
```

```python
#TIME DELTA
df['Date'].max() - df['Date'].min()
```

Filtering by date

```python
filt = (df['Date'] >= '2023') & (df['Date'] < '2024')
df.loc[filt]
```

```python
filt1 = (df['Date'] >= pd.to_datetime('2023')) & (df['Date'] < pd.to_datetime('2024'))
df.loc[filt1]
```

```python
# set index to date
df.set_index('Date', inplace=True)
```

```python
df
```

```python
df.loc['2021']     # returns index/date selection
# df['2024']     column selection
```

Slicing dates

```python
df.index.is_monotonic_increasing
```

```python
df.sort_index(inplace=True)
```

```python
df.loc['2021-01-02' : '2021-08-02']
```

```python
cols = ['Open','High','Low','Price']

df[cols] = df[cols].replace(',', '', regex=True).astype(float)
```

```python
#Avg open for Jan 2023 to Feb 2024
df.loc['2022-01-02' : '2023-02-02']['Open'].mean()
```

Resample data

```python
#resample 1 column
df['High'].resample('W').max()
```

```python
#Resample multiple columns
df.resample('W').agg({'High': 'max', 'Low': 'min', 'Open':'mean'})
```
