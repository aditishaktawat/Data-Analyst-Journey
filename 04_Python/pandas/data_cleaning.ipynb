```python
import pandas as pd
import numpy as np
```

```python
#Dictionary with multiple values in key using list
person = {
    "First_Name" : ["Aditi", "Joe", "Anu", np.nan, None, "Missing"],
    "Last_Name" : ["Shaktawat", "Sin", "Dat",np.nan,np.nan,"N/A"],
    "Email" : ["abc@email.com", "av@email.com", "fgees@email.com","N/A",None, "Missing"],
    "Age": [34,12,67,np.nan, "Missing", None]}
```

```python
df = pd.DataFrame(person)
df
```

Handling Null Values
1. Remove null values

```python
df.dropna()
```

```python
df.dropna(axis='index', how='all')    #if all rows null, only them remove them
```

```python
df.dropna(axis='columns',how='any')   # removes columns which have any value in the column as null
```

```python
df.dropna(axis='columns', how='all')   #only if all values in column are empty
```

```python
# If either 'Last name' or 'email' present , dont remove those values
 #  OR
# Remove rows which dont have email or last name

df.dropna(axis='index',how='any',subset=['Email','Last_Name'])
```

```python
df.isna()
```

2. Remove custom missing values

```python
df.replace('N/A',np.nan,inplace=True)
df.replace('Missing',np.nan,inplace=True)
df
```

3. Filling Null with some different values

```python
df.fillna("Not-Present")
```

```python
df['Age'].fillna(0)
```
